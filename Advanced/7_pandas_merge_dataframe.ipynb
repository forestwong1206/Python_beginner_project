```python
# Import libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading dataset
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data clean up
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df_usa = df[df['job_country']=='United States'].copy()
df_usa['job_posted_month'] = df_usa['job_posted_date'].dt.strftime('%B')
usa_pivot = df_usa.pivot_table(index= 'job_posted_month', columns='job_title_short', aggfunc= 'size')
usa_pivot.reset_index(inplace=True)
usa_pivot['month_no'] = pd.to_datetime(usa_pivot['job_posted_month'], format='%B').dt.month
usa_pivot.sort_values('month_no', inplace = True)
usa_pivot.set_index('job_posted_month', inplace = True)
usa_pivot.drop(columns='month_no', inplace = True)
usa_pivot
```

```python
usa_software_pivot = pd.read_csv('https://lukeb.co/software_csv', index_col = 'job_posted_month')
usa_software_pivot
```

```python
usa_merge = usa_pivot.merge(usa_software_pivot, on='job_posted_month')
usa_merge
```

```python
top_5 = usa_merge.sum().sort_values(ascending=False).head().index.to_list()
top_5
```

```python
usa_merge[top_5].plot(kind='line')
```
