```python
# Import libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading dataset
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data clean up
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df.sample(5)
```

```python
df.index
```

```python
df.index.name = 'job_index'
```

```python
df.sample(5)
```

```python
df.index.dtype
```

```python
df_usa = df[df['job_country'] == 'United States']
df_usa.head(5)
#Now the job_index is messed up
```

```python
#reset the index
df_usa.reset_index(inplace =True)
df_usa.head(5)
```

```python
# now we can use a specific column as the index, for now lets use the job_index
df_usa.set_index('job_index', inplace=True)
df_usa.head(5)
```

sort_index()
* useful when using a pivot table

```python
#Lets say we built a pivot table
median_pivot = df_usa.pivot_table(values= 'salary_year_avg', index='job_title_short', aggfunc=['median','mean','max'])
median_pivot
```

```python
#Let say we sort it by median
median_pivot = median_pivot.sort_values(('median', 'salary_year_avg'), ascending=False)
median_pivot
```

```python
#But then we want to sort it back as alphebetical order by job title
median_pivot.sort_index(inplace=True)
```
