df.pivot_table

```python
# Import libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading dataset
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data clean up
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df.pivot_table(index = 'job_title_short', aggfunc = 'size')
# Size is to count all rows, Count is to count all non-NA values in a specific column
```

```python
df.pivot_table(values = 'salary_year_avg', index = 'job_title_short', columns = 'job_country', aggfunc = 'median')
df.pivot_table(values = 'salary_year_avg', index = 'job_country', columns = 'job_title_short', aggfunc = 'median')
```

```python
top_countries = df['job_country'].value_counts().head(6).index 
top_countries
# we want to rank the countries by the number of job postings, then capture the top 6, then only want the index
```

```python
df_job_country_salary = df.pivot_table(values = 'salary_year_avg', index = 'job_country', columns = 'job_title_short', aggfunc = 'median')

#Now we need top 6 only
df_job_country_salary = df_job_country_salary.loc[top_countries]

#Now we need 3 job roles only
job_titles = ['Data Analyst', 'Data Engineer', 'Data Scientist']
df_job_country_salary =  df_job_country_salary[job_titles]

df_job_country_salary
```

```python
df_job_country_salary.plot(kind='bar')
plt.xlabel('')
plt.ylabel('Median Salary (USD)')
plt.title('Median Salary by Country and Job')
plt.xticks(rotation=45, ha='right')
```
