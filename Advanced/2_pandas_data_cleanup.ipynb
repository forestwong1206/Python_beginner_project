```python
# Import libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading dataset
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data clean up
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
type[df.salary_hour_avg]
```

Fill in none values with median values

```python
median_salary_year_avg = df['salary_year_avg'].median()
```

```python
median_salary_hour_avg = df['salary_hour_avg'].median()
```

```python
df_filled = df

df_filled['salary_year_avg'] = df_filled['salary_year_avg'].fillna(median_salary_year_avg)
df_filled['salary_hour_avg'] = df_filled['salary_hour_avg'].fillna(median_salary_hour_avg)

# Use fillna to fill values with null 
```

```python
df_filled.loc[0:10, ['salary_year_avg', 'salary_hour_avg']]
```

Drop Duplicate values

```python
df_unique = df_filled

df_unique = df_unique.drop_duplicates()

print("Rows removed :", len(df_filled) - len(df_unique))
```

Drop duplicates with same job title and company name

```python
df_unique = df_unique.drop_duplicates(subset = ['job_title', 'company_name'])
print("Rows removed :", len(df_filled) - len(df_unique))
```
