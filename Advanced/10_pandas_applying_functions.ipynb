```python
# Import libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading dataset
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data clean up
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

Let say we want to calculate the expected salary with inflation

```python
df_salary = df[pd.notna(df['salary_year_avg'])].copy()

def projected_salary(salary):
    return salary * 1.03

df_salary['salary_year_inflated'] = df_salary['salary_year_avg'].apply(projected_salary)
# apply simply run this function separately on each item in that column 
# df_salary['salary_year_inflated'] = projected_salary(df_salary['salary_year_avg'])
# df_salary['salary_year_inflated'] = df_salary['salary_year_avg'] * 1.03
# df_salary['salary_year_inflated'] = df_salary['salary_year_avg'].apply(lambda salary: salary * 1.03)



df_salary[['salary_year_inflated', 'salary_year_avg']]

```

Convert skills set back to a list (rather than str)

```python
import ast
```

```python
def clean_list(skill_list):
    if pd.notna(skill_list):
        return ast.literal_eval(skill_list)


df['job_skills'] = df['job_skills'].apply(clean_list)
```

```python
#same as above, with lambda
df['job_skills'] = df['job_skills'].apply(
    lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

## For Rows 
* senior role 5%
* other role 3%

```python
def projected_salary(row):
    if 'Senior' in row['job_title_short']:
        return row['salary_year_avg'] * 1.05
    else:
        return row['salary_year_avg'] * 1.03

df['salary_adjusted'] = df.apply(projected_salary, axis=1)
```

```python
#Lambda version again, note that x can be named anythiing
df['salary_adjusted'] = df.apply(
    lambda x: x['salary_year_avg'] * 1.05
    if 'Senior' in x['job_title_short']
    else x['salary_year_avg'] * 1.03,
    axis=1
)
```
